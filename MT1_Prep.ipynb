#### 1. Normalize the input string by lowercasing and removing non-alphabetic, non-whitespace characters


```python
def normalize_string(s):
    assert type (s) is str

    '''order:
    1. s.lower()
    2. assign to c
    3. if condition
    4. if True, append the c to essential_chars
    5. loop 2-4 for every char
    '''
    essential_chars = [c for c in s.lower() if c.isalpha() or c.isspace()]
    return ''.join(essential_chars)
```

```python
# fxn returns a list of splited words which are only lower-case alphabet only
def get_normalized_words(s):
    assert type(s) is str

    # normalize_string fxn removed ','after perspiciatis
    return normalize_string(s).split()
```

#### 2. a list of sets corresponding to distinct letters in the normalized words
- Returns: list of set: A list of sets, each containing the distinct letters of a normalized word. \
The output itemsets should appear in the same order as their corresponding words in the input text.
The normalization process involves lowercasing the text and removing non-alphabetic, non-whitespace
- results : [{'s', 'e', 'd'}, {'t', 'u'}, {'s', 'i', 't', 'p', 'a', 'e', 'c', 'r'}, {'n', 'e', 'd', 'u'}, {'s', 'o', 'm', 'i', 'n'}]

```python
def make_itemsets_unstructured_text(text):

    return [set(word) for word in get_normalized_words(text)]

# Alternative:
    # 1. normalize text first
    n_text = get_normalized_words(text)

    # 2. loop to get the distinct letters from each word
    itemsets = []
    for word in n_text:
        distinct_letters = set(word)
    # 3. within loop, append each set of distinct_letters to a list called 'itemsets'
        itemsets.append (distinct_letters)
    return itemsets
```

#### 3. the pair_counts default dictionary for all item pairs in the given itemset
- Args:\
 pair_counts (defaultdict): The dictionary to update.\
 itemset (set): The set of items to consider.
- Returns: None - The function updates the pair_counts dictionary in place.
- result : After first_itemset:
{('g', 'o'): 1, ('o', 'g'): 1}
After second_itemset:
{('g', 'o'): 2, ('o', 'g'): 2, ('g', 'd'): 1, ('d', 'g'): 1, ('o', 'd'): 1, ('d', 'o'): 1}
#### combinations(iterable, r)
- returns r-length tuples in sorted order

```python
itemsets = [set('go'), set('dog')]

def update_pair_counts(pair_counts, itemset):
    assert type (pair_counts) is defaultdict
    from collections import defaultdict
    from itertools import combinations

    for (a, b) in combinations(itemset, 2):
        pair_counts[(a, b)] += 1
        pair_counts[(b, a)] += 1
```

#### 4.  the item_counts default dictionary for all items in the itemset
- Args:\
item_counts (defaultdict): The dictionary to update.\
itemset (set): The set of items to consider.
- Returns: None - The function updates the item_counts dictionary in place.
- result: After first_itemset:
{'g': 1, 'o': 1}
After second_itemset:
{'g': 2, 'o': 2, 'd': 1}

```python
from collections import defaultdict
item_counts = defaultdict(int)

def update_item_counts(item_counts, itemset):
    assert type(item_counts) is defaultdict

    for a in itemset: # each item is assigned to a
        item_counts[a] += 1 # read its current cnt, add 1, and store the updated cnt
```

#### 5. the Pairwise association rules by calculating confidence
- conf[(a, b)] = pair_counts[(a, b)] / item_counts[a]\
 =  confidence of the rule a → b
- calculates *how often b appears when a appears*, and stores the result in conf.\
pair_counts[(a, b)] — get the number of times a and b appear together.\
Store the result in conf, using the tuple (a, b) as the dictionary key.

- create Association rules from pair_counts and item_counts
- results (eg) :\
conf(blue fish => red fish) = 0.818\
conf(one fish => red fish) = 0.809\
conf(two fish => red fish) = 0.797



```python
def create_rules_from_counts(pair_counts, item_counts):
    rules = {} # (item_a, item_b) -> conf (item_a => item_b)

    for (a, b) in pair_counts:
        assert a in item_counts

        conf_ab = pair_counts[(a, b)] / item_counts[a]
        rules[(a, b)] = conf_ab

    return rules
```
