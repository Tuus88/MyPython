#### 1. Normalize the input string by lowercasing and removing non-alphabetic, non-whitespace characters


```python
def normalize_string(s):
    assert type (s) is str

    '''order:
    1. s.lower()
    2. assign to c
    3. if condition
    4. if True, append the c to essential_chars
    5. loop 2-4 for every char
    '''
    essential_chars = [c for c in s.lower() if c.isalpha() or c.isspace()]
    return ''.join(essential_chars)
```

#### 3. the pair_counts default dictionary for all item pairs in the given itemset

#### combinations(iterable, r)

```python
itemsets = [set('go'), set('dog')]

def update_pair_counts(pair_counts, itemset):
    assert type (pair_counts) is defaultdict
    from collections import defaultdict
    from itertools import combinations

    for (a, b) in combinations(itemset, 2):
        pair_counts[(a, b)] += 1
        pair_counts[(b, a)] += 1
```

#### 1. the Pairwise association rules by calculating confidence
- conf[(a, b)] = pair_counts[(a, b)] / item_counts[a]\
 =  confidence of the rule a → b
- calculates *how often b appears when a appears*, and stores the result in conf.\
pair_counts[(a, b)] — get the number of times a and b appear together.\
Store the result in conf, using the tuple (a, b) as the dictionary key.

- create Association rules from pair_counts and item_counts
- results (eg) :\
conf(blue fish => red fish) = 0.818\
conf(one fish => red fish) = 0.809\
conf(two fish => red fish) = 0.797



```python
def create_rules_from_counts(pair_counts, item_counts):
    rules = {} # (item_a, item_b) -> conf (item_a => item_b)

    for (a, b) in pair_counts:
        assert a in item_counts

        conf_ab = pair_counts[(a, b)] / item_counts[a]
        rules[(a, b)] = conf_ab

    return rules
```

#### 2. Filter the confidence ules based on a threshold
- remove any rules not exceeding threshold
- returns: A dict containing only the rules with confidence >= threshold.
- filtered_rules.update({rule: conf})\
If rule already exists in filtered_rules, its value is replaced with conf.\
If rule does not exist, a new entry is added.

```python
def filter_rules_by_conf(rules, threshold):
    # order of processing
    # 1. evaluate rules.items()
    # 2. if condition
    # 3. rule:conf - add qualifying pair to the new dict (rule=key, conf=value)
    # 4. repeat for every pair
    # 5. return the completed dict
    return {rule: conf for rule, conf in rules.items() if conf >= threshold}

# Alternative:
    filtered_rules = {}

    for rule, conf in rules.items():
        if conf >= threshold:
            filtered_rules[rule] = conf

    return filtered_rules
```

#### 4. create Assoication rules from a data source
- given itemset_maker(), filtered by confdence threshold & min item count
1. Parse "receipts" into sets of distinct "items".\
The user should be able to define how this is done.
2. Process the itemsets into co-occurrence and occurrence counts.
3. Process the counts into confidence rules.
4. Filter the confidence rules based on a threshold.
5. Filter the confidence rules based on a minimum occurrence count.

```python
def create_rules_from_source(source, itemset_maker, conf_threshold, min_count):

    # 1. initialize dict for pair_counts and item_counts
    pc = defaultdict(int)
    ic = defaultdict(int)

    # 2. use itemset_maker fxn to create a list of itemsets
    make_itemsets_csv = itemset_maker(source)

    for itemset in make_itemsets_csv:
        # Ex 3
        pc = update_pair_counts(pc, itemset)
        # Ex 4
        ic = update_item_counts(ic, itemset)

    # 3. Ex 1 - create Association Rules from pc & ic
    rules = create_rules_from_counts(pc, ic)

    # 4. Ex 2 - Filter Rules based on threshold
    filtered_dict = filter_rules_by_conf(rules, conf_threshold)

    # 5. filter based on min_count
    final_dict = {}
    for pair, conf in filtered_dict.items():
        if ic[pair[0]] >= min_count:
            final_dict[pair] = conf
    return final_dict
```

- z = x.intersection(y)

```python
def common_rules(rules_list):

    result = None
    #print(rules_list)
    # iterate through the rules_list
    for rules in rules_list:
        # if result = None then set to rules.keys()
        if result is None:
            result = set(rules.keys()) #returns all the keys in the dict rules

        # else get intersection
        else:
            result = result.intersection(set(rules.keys()))

    return result
```

#### 4. convert a floating-point value (sign, significand, exponent) tuple in a base to a float
- sign: +/-, signficand: str, exponent: int
- eval_fp(-, 2.G0EBPT, -1, 32)
--> -0.0781387033930514

```python
def eval_fp(sign, significand, exponent, base=2):
    assert sign in ['+', '-'], "Sign bit must be '+' or '-', not '{}'.".format(sign)
    assert type(exponent) is int

    ### BEGIN SOLUTION
    v_sign = 1.0 if sign == '+' else -1.0
    v_significand = eval_strfrac(significand, base)
    return v_sign * v_significand * (base ** exponent)

    # my ver ----------------------------------------------------------------
    num = eval_strfrac(significand, base)
    # to move decimal point to a desired position
    num = num * (base ** exponent)
    if sign == '+':
        return num
    else:
        return -num
```
