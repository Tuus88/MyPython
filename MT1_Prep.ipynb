### zip()
- combine 2 or more iterables (tuples, str, dict, list) into a 1 iterator
- unzip: eg) students, avg = zip(*g)



```python
def build_grade_dicts(grades):

    assignments = grades[0][1:]
    return {L[0]: dict(zip(assignments, [int(g) for g in L[1:]])) for L in grades[1:]}

    # Alternative
    grade_dicts = {} # Empty
    for L in grades[1:]:
        grade_dicts[L[0]] = dict(zip(assignments, [int(g) for g in L[1:]]))
    return grade_dicts
```

#### fmean()
- always converts to float avg
#### dict.**items()**
- returns key, value tuples
- used inside a loop to unpack keys and values \
  eg) for k, v in d.items():

```python
def build_avg_by_student(grades):

    grade_lists = {L[0]:[int(g) for g in L[1:]] for L in grades[1:]}

    from statistics import fmean
    return {n: round(fmean(G),3) for n, G in grade_lists.items()}
```

#### enumerate()
- loop over and have an automatic counter
- eg) for i, (k,v) in enumerate(d.items()):

```python
def build_grade_by_asn(grades):

    assignments = grades[0][1:]

    return {a:[int(L[k+1]) for L in grades[1:]] for k,a in enumerate(assignments)} #int(L[k+1])는 각 a에 해당하는 학생별 score를 loop로 배정

    #alternative
    assignments = grades[0][1:] #assignments = 'Exam 1', 'Exam 2', 'Exam 3'
    grades_by_assignment = {} # Empty dictionary
    for k, a in enumerate(assignments): # k = 0, 1, 2.  a = 'Exam 1', 'Exam 2', 'Exam 3'
        grades_by_assignment[a] = [int(L[k+1]) for L in grades[1:]] #grade[1] == ['Thorny', '100', '90', '80'], L[0+1]=='100', L[0+2]=='88'
    return grades_by_assignment
```

- lambda item: (-item[1], item[0])\
  -item[1] ↘ descending order \
  item[0] if 2 values are the same value, ascending order.



```python
def get_ranked_students(grades):

    avg_grades_by_student = build_avg_by_student(grades)
    return [k for k,v in sorted(avg_grades_by_student.items(), key=lambda item: (-item[1],item[0]))]
```
