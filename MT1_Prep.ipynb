### zip()
- combine 2 or more iterables (tuples, str, dict, list) into a 1 iterator
- unzip: eg) students, avg = zip(*g)



```python
def build_grade_dicts(grades):

    assignments = grades[0][1:]
    return {L[0]: dict(zip(assignments, [int(g) for g in L[1:]])) for L in grades[1:]}

    # Alternative
    grade_dicts = {} # Empty
    for L in grades[1:]:
        grade_dicts[L[0]] = dict(zip(assignments, [int(g) for g in L[1:]]))
    return grade_dicts
```

#### fmean()
- always converts to float avg
#### dict.**items()**
- returns key, value tuples
- used inside a loop to unpack keys and values \
  eg) for k, v in d.items():

```python
def build_avg_by_student(grades):

    grade_lists = {L[0]:[int(g) for g in L[1:]] for L in grades[1:]}

    from statistics import fmean
    return {n: round(fmean(G),3) for n, G in grade_lists.items()}
```
